```python
from pocaduck import Query, StorageConfig

cfg = StorageConfig(base_path="/groups/troidl/troidllab/shape_reasoning/liconn_axons_cluster_v3/points")

# Create a query object
query = Query(storage_config=cfg) # path to folder where nps output is stored

# Get all points for a label (aggregated across all blocks)
points = query.get_points(label=104707)
print(f"Retrieved {points.shape[0]} points for label {104707}")
print(points.shape)

# Close the query connection when done
query.close()
```
